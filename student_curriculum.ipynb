```python
curriculum_stages = [
    {"data": "visdrone_stage1_easy.yaml", "epochs": 20},
    {"data": "visdrone_stage2_med.yaml",  "epochs": 50},
    {"data": "visdrone_stage3_full.yaml", "epochs": 130}
    # {"data": "visdrone_stage1_easy.yaml", "epochs": 1},
    # {"data": "visdrone_stage2_med.yaml",  "epochs": 1},
    # {"data": "visdrone_stage3_full.yaml", "epochs": 1}
]

current_weights = "mobilenetv4-yolo26m.yaml" 

for stage_idx, stage_cfg in enumerate(curriculum_stages):
    stage_num = stage_idx + 1
    model = YOLO(current_weights)

    model.train(
        data=stage_cfg["data"],
        epochs=stage_cfg["epochs"],
        batch=16,
        imgsz=1024,
        amp=True,
        patience=50,
        project="student_curriculum",
        name=f"stage_{stage_num}",
    )

    current_weights = f"ultralytics/runs/detect/student_curriculum/stage_{stage_num}/weights/best.pt"
```

```python
best_model = YOLO("ultralytics/runs/detect/student_curriculum/stage_3/weights/best.pt")
best_model.val(data="ultralytics/cfg/datasets/VisDrone.yaml", split="test")
```
